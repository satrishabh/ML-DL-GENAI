```python
from langchain_core.prompts import PromptTemplate
from langchain_classic.chains import LLMChain, SequentialChain
from langchain_classic.memory import ConversationBufferMemory
```

```python
title_prompt=PromptTemplate(
    input_variables=["topic"],
    template="write me a blog title about {topic}"
)

title_chain=LLMChain(llm=LLM, prompt=title_prompt, output_key="title", verbose=True)# output of first chain
```

```python
#chain 2
outline_prompt=PromptTemplate(
    input_variables=["title"],
    template="generate a stucture outline for  a blog with title {title}"
)

outline_chain=LLMChain(llm=LLM, prompt=outline_prompt, output_key="outline", verbose=True)
```

```python
#chain 3
blog_prompt=PromptTemplate(
    input_variables=["title", "outline"],
    template="write a dtailed blog with title {title} and outline {outline}"
)

blog_chain=LLMChain(llm=LLM, prompt=blog_prompt,output_key="blog", verbose=True)
```

```python
summary_prompt=PromptTemplate(
    input_variables=["blog"],
    template="Summarization the following blog in 3-4 sentences {blog}"
)

summary_chain=LLMChain(llm=LLM, prompt=summary_prompt, output_key="summary", verbose=True)
#
```

```python
sequential_workflow=SequentialChain(
    chains=[title_chain, outline_chain, blog_chain, summary_chain],
    input_variables=["topic"],
    output_variables=["title", "outline", "blog", "summary"],
    verbose=True
)
```

```python
result=sequential_workflow({"topic":"Artificial Intelligence driven cyber security in banking domain"})
```
