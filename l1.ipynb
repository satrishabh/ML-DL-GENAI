```python
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv("/content/fraud_transactions_synthetic.csv")
```

```python
data.head()
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
x=data.drop(['is_fraud'],axis=1)
y=data['is_fraud']
```

```python
x.head()
```

```python
y.head()
```

```python
data.shape
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25)
```

```python
x_test.head()
```

```python
x_test.shape
```

```python
x_train.shape[1]
```

```python
model=keras.Sequential([
    layers.Input(shape=(x_train.shape[1],),name='input_layer'), # input has 10 neurons
    layers.Dense(16,activation='relu',name='hidden_layer_1'),# 16 neuron
    layers.Dense(32,activation='relu',name='hidden_layer_2'),# 32 neuron
    layers.Dense(1,activation='sigmoid',name='output_layer')# output_layer has just 1 neuron
])
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(x_train,y_train,epochs=10,batch_size=32)
```
